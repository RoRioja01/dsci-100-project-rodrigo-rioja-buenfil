This scatterplot shows the relationship between conference wins and tournament wins. We can see that having more conference wins does not always mean winning more games in the NCAA tournament, since many teams with a high number of conference wins still won zero tournament games. This suggests that conference wins alone might not be enough to predict tournament wins, but they could still be useful when combined with conference losses in our regression model.

```r
wins_plot <- ggplot(wbdata_select, aes(x = conf_wins, y = tourney_wins)) +
    geom_point(alpha = 0.2) +
    labs(title = "Conference Wins vs Tournament Wins", x = "Conference Wins", y = "Tournament Wins")

wins_plot
```
